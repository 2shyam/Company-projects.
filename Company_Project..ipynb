```python
from nltk.corpus import gutenberg 
files=gutenberg.fileids()
print(files)
```

```python
raw_text = gutenberg.raw('austen-emma.txt')
print(raw_text[:500])
```

```python
words = gutenberg.words('austen-emma.txt')
print(words[:50])
```

```python
all_text =""
for file_id in files:
    all_text+= gutenberg.raw(file_id) + "\n\n"
print("total characters in dataset: ", len(all_text))    
```

```python
print(all_text[:1000])
```

```python
text=all_text.lower().replace("\n"," ")
print(text[:500])
```

```python
import re
from nltk.tokenize import word_tokenize 
nltk.download('punkt')

clean_text =re.sub(r'[^a-zA-Z\s]',"",text)
tokens=word_tokenize(clean_text)
print(tokens[:50])
```

```python
len(tokens)
```

```python
import pandas as pd 
df = pd.DataFrame(tokens,columns=["word"])
df.head(10)
```

```python
df.shape
```

```python
sentences = gutenberg.sents()
df_sentences = pd.DataFrame({"sentence" : [" ".join(sent) for sent in sentences]})
df_sentences.head(10)
len(df_sentences)
```

```python
pairs =[]
for i in range(len(df_sentences)-3):
    prompt = df_sentences.iloc[i]["sentence"]
    continuation = "".join(df_sentences.iloc[i+1 : i +4]["sentence"])
    pairs.append({"prompt": prompt,"continuation":continuation})
print("total pairs: ", len(pairs))
```

```python
pairs[0]
```

```python
###### Optional work for me just ######
df_pairs = pd.DataFrame(pairs)
df_pairs.to_csv("gutenberg_pairs.csv",index=False)
```

```python
from transformers import AutoTokenizer,AutoModelForCausalLM

model_name = "gpt2"
tokenizer = AutoTokenizer.from_pretrained(model_name)
tokenizer.pad_token = tokenizer.eos_token  # GPT2 doesn’t have pad token

model = AutoModelForCausalLM.from_pretrained(model_name)
```

Prepare Dataset For Training

```python
from datasets import Dataset
import datasets
datasets.config.TF_AVAILABLE = False 
datasets.config.TORCH_AVAILABLE = True
dataset = Dataset.from_pandas(df_pairs)
def tokenize_function(example):
    text = example["prompt"] + " <CONT> " + example["continuation"]
    tokenized = tokenizer(text,truncation=True,padding="max_length",max_length=256)
    
    tokenized["labels"] = tokenized["input_ids"].copy()
    return tokenized

tokenized_dataset = dataset.map(tokenize_function,batched=False)
```

```python
from transformers import Trainer,TrainingArguments
print("Trainer works now")
```

Train The Model

```python
from transformers import Trainer,TrainingArguments

training_args = TrainingArguments(output_dir="./gpt2-finetuned",evaluation_strategy="steps",learning_rate=5e-5,per_device_train_batch_size=2,weight_decay=0.01,save_steps=500,logging_steps=100,push_to_hub=False)
trainer = Trainer(model=model,args=training_args,train_dataset=tokenized_dataset,tokenizer=tokenizer)
trainer.train()
```

Save the model

```python
trainer.save_model("./gpt2-finetuned")
tokenizer.save_pretrained("./gpt2-finetuned")
```

Test the model

```python
from transformers import pipeline

generator = pipeline("text-generation", model="./gpt2-finetuned", tokenizer=tokenizer)

prompt = "The young prince entered the dark forest."
output = generator(prompt, max_new_tokens=100, do_sample=True, top_p=0.9, temperature=0.8)

print(output[0]["generated_text"])
```
